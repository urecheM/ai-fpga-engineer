```python
import json
lb = json.load(open('results/leaderboards/baseline-v1_leaderboard.json'))
import matplotlib.pyplot as plt
rows = lb['overall']
plt.bar([r['model'] for r in rows], [r['pass_rate'] for r in rows])
plt.ylabel('pass rate'); plt.xticks(rotation=20); plt.title('baseline-v1'); plt.show()
```
