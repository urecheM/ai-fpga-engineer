```python
import json
from hdleval.leaderboard.aggregate import wilson_interval
lb = json.load(open('results/leaderboards/baseline-v1_leaderboard.json'))
for r in lb['overall']:
    print(f"{r['model']:16s} pass={r['pass_rate']:.3f} CI={r['pass_ci95']}")
```

## Interpretation

Non-overlapping confidence intervals separate the strongest and weakest configurations; pass rate declines with objective difficulty, concentrated on hard communication-protocol benchmarks.

```python
import matplotlib.pyplot as plt
rows = lb['overall']
xs = [r['model'] for r in rows]
ys = [r['pass_rate'] for r in rows]
err = [[y-r['pass_ci95'][0] for y,r in zip(ys,rows)],
       [r['pass_ci95'][1]-y for y,r in zip(ys,rows)]]
plt.errorbar(xs, ys, yerr=err, fmt='o'); plt.xticks(rotation=20)
plt.ylabel('pass rate'); plt.title('baseline-v1 with 95% CI'); plt.show()
```
